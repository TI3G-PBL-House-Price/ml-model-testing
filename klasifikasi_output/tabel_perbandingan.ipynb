### Import Library

```python
import os
import pandas as pd
import joblib
```

### Memuat hasil evaluasi 3 model

```python
hasil_knn = joblib.load('../klasifikasi_model/hasil_knn.pkl')
hasil_svm = joblib.load('../klasifikasi_model/hasil_svm.pkl')
hasil_nb  = joblib.load('../klasifikasi_model/hasil_nb.pkl')

daftar_hasil = [hasil_knn, hasil_svm, hasil_nb]
```

### Tabel Perbandingan

```python
data_komparasi = []

for h in daftar_hasil:
    data_komparasi.append({
        'Model': h['Model'],
        'Accuracy': h['Accuracy'],
        'Accuracy (%)': round(h['Accuracy'] * 100, 2),
        'Precision': h['Precision'],
        'Recall': h['Recall'],
        'F1-Score': h['F1-Score'],
        'Waktu Eksekusi (Detik)': h['Waktu (Detik)']
    })

df_perbandingan = pd.DataFrame(data_komparasi)

print("=== TABEL PERBANDINGAN PERFORMA MODEL ===")
display(df_perbandingan)
```

### Urutan Model Berdasarkan Accuracy Terbaik

```python
df_urut_accuracy = df_perbandingan[['Model', 'Accuracy', 'Accuracy (%)']].sort_values(
    by='Accuracy', ascending=False
).reset_index(drop=True)
df_urut_accuracy.index += 1 

print("\n=== URUTAN MODEL DENGAN ACCURACY TERBAIK ===")
display(df_urut_accuracy)
```

### Urutan Model Berdasarkan Precision Terbaik

```python
df_urut_precision = df_perbandingan[['Model', 'Precision']].sort_values(
    by='Precision', ascending=False
).reset_index(drop=True)
df_urut_precision.index += 1

print("\n=== URUTAN MODEL DENGAN PRECISION TERBAIK ===")
display(df_urut_precision)
```

### Urutan Model Berdasarkan Recall Terbaik

```python
df_urut_recall = df_perbandingan[['Model', 'Recall']].sort_values(
    by='Recall', ascending=False
).reset_index(drop=True)
df_urut_recall.index += 1

print("\n=== URUTAN MODEL DENGAN RECALL TERBAIK ===")
display(df_urut_recall)
```

### Urutan Model Berdasarkan F1-Score Terbaik

```python
df_urut_f1 = df_perbandingan[['Model', 'F1-Score']].sort_values(
    by='F1-Score', ascending=False
).reset_index(drop=True)
df_urut_f1.index += 1

print("\n=== URUTAN MODEL DENGAN F1-SCORE TERBAIK ===")
display(df_urut_f1)
```

### Urutan Model Berdasarkan Waktu Eksekusi Terbaik (Tercepat)

```python
df_urut_waktu = df_perbandingan[['Model', 'Waktu Eksekusi (Detik)']].sort_values(
    by='Waktu Eksekusi (Detik)', ascending=True
).reset_index(drop=True)
df_urut_waktu.index += 1

print("\n=== 5. URUTAN MODEL DENGAN WAKTU EKSEKUSI TERCEPAT ===")
display(df_urut_waktu)
```

### Menyimpan Tabel Utama untuk Visualisasi

```python
joblib.dump(df_perbandingan, 'df_perbandingan.pkl')
print("\nData perbandingan berhasil disimpan ke 'klasifikasi_output/df_perbandingan.pkl'!")
```
