### Import Library

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.metrics import confusion_matrix
```

### Memuat tabel perbandingan

```python
df_perbandingan = joblib.load('df_perbandingan.pkl')
print("Data perbandingan berhasil dimuat!")
display(df_perbandingan)
```

### Visualisasi Gabungan 4 Metrik Utama

```python
metrik_cols = ['Accuracy', 'Precision', 'Recall', 'F1-Score']
df_melted = df_perbandingan.melt(
    id_vars='Model', 
    value_vars=metrik_cols, 
    var_name='Metrik', 
    value_name='Skor'
)

plt.figure(figsize=(10, 5))
ax = sns.barplot(data=df_melted, x='Model', y='Skor', hue='Metrik', palette='Set2', edgecolor='black')

plt.title('Perbandingan Metrik Evaluasi Antar Model (Accuracy, Precision, Recall, F1-Score)', fontsize=13, fontweight='bold')
plt.xlabel('Model', fontsize=11)
plt.ylabel('Skor (0.0 - 1.0)', fontsize=11)
plt.ylim(0, 1.15)
plt.legend(title='Metrik Evaluasi', loc='upper right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Menampilkan angka di atas masing-masing batang
for container in ax.containers:
    ax.bar_label(container, fmt='%.3f', padding=3, fontsize=9)

plt.tight_layout()
plt.show()
```

### Visualisasi Perbandingan Accuracy per Model

```python
df_urut_acc = df_perbandingan.sort_values(by='Accuracy (%)', ascending=False)

plt.figure(figsize=(7, 4))
ax = sns.barplot(
    data=df_urut_acc, x='Model', y='Accuracy (%)', 
    hue='Model', palette='Blues_r', edgecolor='black', legend=False
)

plt.title('Perbandingan Tingkat Akurasi Model (%)', fontsize=12, fontweight='bold')
plt.xlabel('Model')
plt.ylabel('Accuracy (%)')
plt.ylim(0, 110)
plt.grid(axis='y', linestyle='--', alpha=0.7)

for container in ax.containers:
    ax.bar_label(container, fmt='%.2f%%', padding=3, fontweight='bold')

plt.tight_layout()
plt.show()
```

### Visualisasi Perbandingan Precision per Model

```python
df_urut_prec = df_perbandingan.sort_values(by='Precision', ascending=False)

plt.figure(figsize=(7, 4))
ax = sns.barplot(
    data=df_urut_prec, x='Model', y='Precision', 
    hue='Model', palette='Greens_r', edgecolor='black', legend=False
)

plt.title('Perbandingan Tingkat Precision Model', fontsize=12, fontweight='bold')
plt.xlabel('Model')
plt.ylabel('Precision')
plt.ylim(0, 1.1)
plt.grid(axis='y', linestyle='--', alpha=0.7)

for container in ax.containers:
    ax.bar_label(container, fmt='%.4f', padding=3, fontweight='bold')

plt.tight_layout()
plt.show()
```

### Visualisasi Perbandingan Recall per Model

```python
df_urut_rec = df_perbandingan.sort_values(by='Recall', ascending=False)

plt.figure(figsize=(7, 4))
ax = sns.barplot(
    data=df_urut_rec, x='Model', y='Recall', 
    hue='Model', palette='Purples_r', edgecolor='black', legend=False
)

plt.title('Perbandingan Tingkat Recall Model', fontsize=12, fontweight='bold')
plt.xlabel('Model')
plt.ylabel('Recall')
plt.ylim(0, 1.1)
plt.grid(axis='y', linestyle='--', alpha=0.7)

for container in ax.containers:
    ax.bar_label(container, fmt='%.4f', padding=3, fontweight='bold')

plt.tight_layout()
plt.show()
```

### Visualisasi Perbandingan F1-Score per Model

```python
df_urut_f1 = df_perbandingan.sort_values(by='F1-Score', ascending=False)

plt.figure(figsize=(7, 4))
ax = sns.barplot(
    data=df_urut_f1, x='Model', y='F1-Score', 
    hue='Model', palette='Reds_r', edgecolor='black', legend=False
)

plt.title('Perbandingan Tingkat F1-Score Model', fontsize=12, fontweight='bold')
plt.xlabel('Model')
plt.ylabel('F1-Score')
plt.ylim(0, 1.1)
plt.grid(axis='y', linestyle='--', alpha=0.7)

for container in ax.containers:
    ax.bar_label(container, fmt='%.4f', padding=3, fontweight='bold')

plt.tight_layout()
plt.show()
```

### Visualisasi Perbandingan Waktu Eksekusi (Detik)

```python
df_urut_waktu = df_perbandingan.sort_values(by='Waktu Eksekusi (Detik)', ascending=True)

plt.figure(figsize=(7, 4))
ax = sns.barplot(
    data=df_urut_waktu, x='Model', y='Waktu Eksekusi (Detik)', 
    hue='Model', palette='Oranges_r', edgecolor='black', legend=False
)

plt.title('Perbandingan Waktu Eksekusi Model (Lebih Kecil = Lebih Cepat)', fontsize=12, fontweight='bold')
plt.xlabel('Model')
plt.ylabel('Waktu (Detik)')
plt.ylim(0, 1.1)
plt.grid(axis='y', linestyle='--', alpha=0.7)

for container in ax.containers:
    ax.bar_label(container, fmt='%.4f dtk', padding=3, fontweight='bold')

plt.tight_layout()
plt.show()
```
