### Import Library

```python
import pandas as pd
import joblib
```

### Memuat hasil evaluasi 3 model Regresi

```python
hasil_knn = joblib.load('../regresi_model/hasil_knn.pkl')
hasil_rf  = joblib.load('../regresi_model/hasil_rf.pkl')
hasil_svm = joblib.load('../regresi_model/hasil_svm.pkl')

daftar_hasil = [hasil_knn, hasil_rf, hasil_svm]
print("Seluruh data hasil evaluasi regresi berhasil dimuat!")
```

### Tabel Utama Perbandingan Performa Model Regresi

```python
data_komparasi = []

for h in daftar_hasil:
    data_komparasi.append({
        'Model': h['Model'],
        'R2 Score': h['R2'],
        'R2 (%)': round(h['R2'] * 100, 2),
        'MAE': h['MAE'],
        'RMSE': h['RMSE'],
        'MAPE (%)': round(h['MAPE (%)'], 2),
        'Waktu Eksekusi (Detik)': h['Waktu (Detik)']
    })

df_perbandingan = pd.DataFrame(data_komparasi)

print("=== TABEL UTAMA PERBANDINGAN PERFORMA MODEL REGRESI ===")
display(df_perbandingan)
```

### Urutan Model Berdasarkan R2 Score Terbaik (Tertinggi)

```python
df_urut_r2 = df_perbandingan[['Model', 'R2 Score', 'R2 (%)']].sort_values(
    by='R2 Score', ascending=False
).reset_index(drop=True)
df_urut_r2.index += 1

print("=== URUTAN MODEL DENGAN R2 SCORE TERBAIK ===")
display(df_urut_r2)
```

### Urutan Model Berdasarkan MAE Terbaik (Terkecil)

```python
df_urut_mae = df_perbandingan[['Model', 'MAE']].sort_values(
    by='MAE', ascending=True
).reset_index(drop=True)
df_urut_mae.index += 1

print("=== URUTAN MODEL DENGAN MAE TERBAIK (TERKECIL) ===")
display(df_urut_mae)
```

### Urutan Model Berdasarkan RMSE Terbaik (Terkecil)

```python
df_urut_rmse = df_perbandingan[['Model', 'RMSE']].sort_values(
    by='RMSE', ascending=True
).reset_index(drop=True)
df_urut_rmse.index += 1

print("=== URUTAN MODEL DENGAN RMSE TERBAIK (TERKECIL) ===")
display(df_urut_rmse)
```

### Urutan Model Berdasarkan MAPE Terbaik (Terkecil)

```python
df_urut_mape = df_perbandingan[['Model', 'MAPE (%)']].sort_values(
    by='MAPE (%)', ascending=True
).reset_index(drop=True)
df_urut_mape.index += 1

print("=== URUTAN MODEL DENGAN MAPE TERBAIK (TERKECIL) ===")
display(df_urut_mape)
```

### Urutan Model Berdasarkan Waktu Eksekusi Terbaik (Tercepat)

```python
df_urut_waktu = df_perbandingan[['Model', 'Waktu Eksekusi (Detik)']].sort_values(
    by='Waktu Eksekusi (Detik)', ascending=True
).reset_index(drop=True)
df_urut_waktu.index += 1

print("=== URUTAN MODEL DENGAN WAKTU EKSEKUSI TERCEPAT ===")
display(df_urut_waktu)
```

### Menyimpan Tabel Utama untuk Visualisasi Regresi

```python
joblib.dump(df_perbandingan, 'df_perbandingan.pkl')
print("Data perbandingan regresi berhasil disimpan ke 'regresi_output/df_perbandingan.pkl'!")
```
