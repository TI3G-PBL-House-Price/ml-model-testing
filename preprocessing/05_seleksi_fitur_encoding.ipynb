### Import Library

```python
import pandas as pd
import joblib
from sklearn.preprocessing import LabelEncoder
```

### Memuat data hasil Feature Engineering

```python
df = joblib.load('../dataset/step_04_engineered.pkl')
print("Data berhasil dimuat! Ukuran data:", df.shape)
```

### Seleksi Fitur

```python
df = df[
    [
        'price',
        'district',
        'city',
        'bed_rooms',
        'bath_rooms',
        'carport',
        'land_area',
        'building_area',
        'total_rooms',
        'building_ratio',
        'remaining_land',
        'total_facilities',
        'total_area'
    ]
]

display(df.head())
```

### Verifikasi Jumlah Kategori

```python
print("Jumlah District :", df['district'].nunique())
print("Jumlah City :", df['city'].nunique())
```

### Encoding District

```python
le_district = LabelEncoder()
df['district'] = le_district.fit_transform(df['district'].astype(str))
```

### Encoding City

```python
le_city = LabelEncoder()
df['city'] = le_city.fit_transform(df['city'].astype(str))
```

### Verifikasi Hasil Encoding

```python
display(df.head())
```

### Menyimpan Data Hasil Seleksi Fitur & Encoding

```python
joblib.dump(df, '../dataset/step_05_selected.pkl')
print("\nData berhasil disimpan ke '../dataset/step_05_selected.pkl'!")
```
