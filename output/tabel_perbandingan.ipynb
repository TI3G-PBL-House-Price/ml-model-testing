### Import Library & Memuat Hasil Evaluasi Mode

```python
import pandas as pd
import joblib

# Memuat hasil evaluasi ketiga model dari folder model/
hasil_lr  = joblib.load('../model/hasil_lr.pkl')
hasil_rf  = joblib.load('../model/hasil_rf.pkl')
hasil_gbt = joblib.load('../model/hasil_gbt.pkl')

daftar_model = [hasil_lr, hasil_rf, hasil_gbt]

rekap_global = []
for h in daftar_model:
    m_tr = h['Training']
    m_te = h['Testing']
    rekap_global.append({
        'Model': h['Model'],
        'RMSE (Training)': m_tr['Root mean squared error'],
        'RMSE (Testing)': m_te['Root mean squared error'],
        'MSE (Training)': m_tr['Mean squared error'],
        'MSE (Testing)': m_te['Mean squared error'],
        'MAE (Training)': m_tr['Mean absolute error'],
        'MAE (Testing)': m_te['Mean absolute error'],
        'MAPE (Training)': m_tr['Mean absolute percentage error'],
        'MAPE (Testing)': m_te['Mean absolute percentage error'],
        'R2 (Training)': m_tr['R2'],
        'R2 (Testing)': m_te['R2'],
        'Akurasi R2 Testing (%)': round(m_te['R2'] * 100, 1),
        'Adjusted R2 (Testing)': m_te['Adjusted R2']
    })

df_global = pd.DataFrame(rekap_global)
print("Data hasil evaluasi berhasil dimuat!")
```

### Tabel Global Perbandingan Akhir Ketiga Model

```python
print("TABEL GLOBAL PERBANDINGAN AKHIR KETIGA MODEL:")
display(df_global)
```

### Urutan Model Terbaik Berdasarkan RMSE

```python
df_rmse = df_global[['Model', 'RMSE (Testing)', 'RMSE (Training)']].sort_values(
    by='RMSE (Testing)', ascending=True
).reset_index(drop=True)
df_rmse.index += 1

print("URUTAN MODEL TERBAIK BERDASARKAN RMSE TESTING TERKECIL:")
display(df_rmse)
```

### Urutan Model Terbaik Berdasarkan MSE

```python
df_mse = df_global[['Model', 'MSE (Testing)', 'MSE (Training)']].sort_values(
    by='MSE (Testing)', ascending=True
).reset_index(drop=True)
df_mse.index += 1

print("URUTAN MODEL TERBAIK BERDASARKAN MSE TESTING TERKECIL:")
display(df_mse)
```

### Urutan Model Terbaik Berdasarkan MAE

```python
df_mae = df_global[['Model', 'MAE (Testing)', 'MAE (Training)']].sort_values(
    by='MAE (Testing)', ascending=True
).reset_index(drop=True)
df_mae.index += 1

print("URUTAN MODEL TERBAIK BERDASARKAN MAE TESTING TERKECIL:")
display(df_mae)
```

### Urutan Model Terbaik Berdasarkan MAPE

```python
df_mape = df_global[['Model', 'MAPE (Testing)', 'MAPE (Training)']].sort_values(
    by='MAPE (Testing)', ascending=True
).reset_index(drop=True)
df_mape.index += 1
print("URUTAN MODEL TERBAIK BERDASARKAN MAPE TESTING TERKECIL ===")
display(df_mape)
```
