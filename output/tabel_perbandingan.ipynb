```python
import pandas as pd
import joblib

hasil_lr  = joblib.load('../model/hasil_lr.pkl')
hasil_rf  = joblib.load('../model/hasil_rf.pkl')
hasil_gbt = joblib.load('../model/hasil_gbt.pkl')

daftar_model = [hasil_lr, hasil_rf, hasil_gbt]

rekap_testing = []
for h in daftar_model:
    m_tr = h['Training']
    m_te = h['Testing']
    rekap_testing.append({
        'Model': h['Model'],
        'RMSE (Training)': m_tr['Root mean squared error'],
        'RMSE (Testing)': m_te['Root mean squared error'],
        'MAE (Testing)': m_te['Mean absolute error'],
        'MSE (Testing)': m_te['Mean squared error'],
        'MAPE (Testing)': m_te['Mean absolute percentage error'],
        'R2 (Testing)': m_te['R2'],
        'Akurasi R2 (%)': round(m_te['R2'] * 100, 1),
        'Adjusted R2': m_te['Adjusted R2']
    })

df_rekap = pd.DataFrame(rekap_testing)
print("TABEL PERBANDINGAN AKHIR KETIGA MODEL: ")
display(df_rekap)

print("\nURUTAN MODEL TERBAIK BERDASARKAN RMSE TESTING TERKECIL:")
display(df_rekap.sort_values(by='RMSE (Testing)', ascending=True).reset_index(drop=True))

joblib.dump(df_rekap, 'df_rekap.pkl')
print("Data tabel perbandingan berhasil disimpan ke 'output/df_rekap.pkl'!")
```
