```python
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

df_rekap = joblib.load('df_rekap.pkl')

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

# Grafik Perbandingan RMSE Testing (Semakin kecil semakin baik)
df_urut_rmse = df_rekap.sort_values(by='RMSE (Testing)', ascending=True)
sns.barplot(
    data=df_urut_rmse, x='Model', y='RMSE (Testing)',
    hue='Model', palette='Reds_r', edgecolor='black', legend=False, ax=axes[0]
)
axes[0].set_title('Perbandingan Error RMSE Testing (Terkecil = Terbaik)', fontweight='bold')
axes[0].set_ylim(0, df_urut_rmse['RMSE (Testing)'].max() * 1.2)
axes[0].tick_params(axis='x', rotation=15)
axes[0].grid(axis='y', linestyle='--', alpha=0.7)
for container in axes[0].containers:
    axes[0].bar_label(container, fmt='%.3f', padding=3, fontweight='bold')

# Grafik Perbandingan Akurasi R2 (%) Testing (Semakin besar semakin baik)
df_urut_r2 = df_rekap.sort_values(by='Akurasi R2 (%)', ascending=False)
sns.barplot(
    data=df_urut_r2, x='Model', y='Akurasi R2 (%)',
    hue='Model', palette='Blues_r', edgecolor='black', legend=False, ax=axes[1]
)
axes[1].set_title('Perbandingan Akurasi R2 Testing (%) (Terbesar = Terbaik)', fontweight='bold')
axes[1].set_ylim(0, 105)
axes[1].tick_params(axis='x', rotation=15)
axes[1].grid(axis='y', linestyle='--', alpha=0.7)
for container in axes[1].containers:
    axes[1].bar_label(container, fmt='%.1f%%', padding=3, fontweight='bold')

plt.tight_layout()
plt.show()
```
