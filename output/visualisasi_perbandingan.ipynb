### Import Library & Memuat Data Rekap

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

# Memuat data rekapitulasi evaluasi yang telah disimpan sebelumnya dari folder output/
df_rekap = joblib.load('df_rekap.pkl')

print("Data rekapitulasi berhasil dimuat untuk divisualisasikan!")
display(df_rekap[['Model', 'RMSE (Testing)', 'MSE (Testing)', 'MAE (Testing)', 'MAPE (Testing)', 'R2 (Testing)']])
```

### Visualisasi Bar Chart RMSE (Testing)

```python
plt.figure(figsize=(8, 4.5))
df_urut_rmse = df_rekap.sort_values(by='RMSE (Testing)', ascending=True)

ax = sns.barplot(
    data=df_urut_rmse, x='Model', y='RMSE (Testing)',
    hue='Model', palette='Reds_r', edgecolor='black', legend=False
)

plt.title('Perbandingan Root Mean Squared Error (RMSE) Testing', fontweight='bold', fontsize=12)
plt.xlabel('Model Regresi', fontweight='bold')
plt.ylabel('Nilai RMSE (Terkecil = Terbaik)', fontweight='bold')
plt.ylim(0, df_urut_rmse['RMSE (Testing)'].max() * 1.2)
plt.grid(axis='y', linestyle='--', alpha=0.7)

for container in ax.containers:
    ax.bar_label(container, fmt='%.3f', padding=3, fontweight='bold')

plt.tight_layout()
plt.show()
```

### Visualisasi Bar Chart R2 Testing (%)

```python
plt.figure(figsize=(8, 4.5))
df_urut_r2 = df_rekap.sort_values(by='Akurasi R2 Testing (%)', ascending=False)

ax = sns.barplot(
    data=df_urut_r2, x='Model', y='Akurasi R2 Testing (%)',
    hue='Model', palette='Blues_r', edgecolor='black', legend=False
)

plt.title('Perbandingan Akurasi R-Squared (R2) Testing (%)', fontweight='bold', fontsize=12)
plt.xlabel('Model Regresi', fontweight='bold')
plt.ylabel('Akurasi R2 (%) (Terbesar = Terbaik)', fontweight='bold')
plt.ylim(0, 105)
plt.grid(axis='y', linestyle='--', alpha=0.7)

for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%', padding=3, fontweight='bold')

plt.tight_layout()
plt.show()
```
